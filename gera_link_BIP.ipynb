```python
import pandas as pd
import re
import os

# ==============================
# LINK MODELO DA PETROBRAS
# ==============================

link_modelo = (
    "https://bancodeimagens.petrobras.com.br/fotoweb/archives/"
    "5016-Refino/Folder%203/Dig47498.jpg.info"
    "#c=%2Ffotoweb%2Farchives%2F5016-Refino%2F"
)

# ==============================
# CAMINHO DO ARQUIVO TXT
# ==============================

arquivo_txt = r"C:\Users\GG88\OneDrive - PETROBRAS\Área de Trabalho\ebooks xp\PROJETO APLICADO XP\BASE DE DADOS\DIg fotos\fotos_exploração_e_produção.txt"

# ==============================
# EXPRESSÃO REGULAR PARA DigXXXXX.jpg
# ==============================

padrao_dig = re.compile(r"Dig\d+\.jpg")

dados = []

# ==============================
# VERIFICAR SE O ARQUIVO EXISTE
# ==============================

if os.path.exists(arquivo_txt):

    with open(arquivo_txt, "r", encoding="utf-8") as f:

        for linha in f:

            codigo = linha.strip()

            if codigo:

                novo_link = padrao_dig.sub(codigo, link_modelo)

                dados.append({
                    "codigo": codigo,
                    "link": novo_link
                })

    # ==============================
    # CRIAR DATAFRAME
    # ==============================

    df = pd.DataFrame(dados)

    # ==============================
    # SALVAR EXCEL
    # ==============================

    arquivo_excel = r"C:\Users\GG88\OneDrive - PETROBRAS\Área de Trabalho\links_refino.xlsx"

    df.to_excel(arquivo_excel, index=False)

    print("✅ Links gerados com sucesso!")
    print(f"Arquivo salvo em: {arquivo_excel}")

else:

    print("❌ Arquivo TXT não encontrado.")
    print(arquivo_txt)
```
