```python
import pandas as pd
import requests
import os

# Caminho do CSV e da pasta de imagens
csv_path = r"C:\Users\GG88\OneDrive - PETROBRAS\Área de Trabalho\Gabriel_testes\BANCO_DE_IMAGEM_TESTE\link-caminho_foto.csv"
base_dir = r"C:\Users\GG88\OneDrive - PETROBRAS\Área de Trabalho\Gabriel_testes\BANCO_DE_IMAGEM_TESTE\414_imagens"

# Leia o CSV
df = pd.read_csv(csv_path, header=None)

# Parâmetros da API
base_url = "https://apit.petrobras.com.br/ia/visao/v1/azure-ai-vision/vision/v3.2/describe"
headers = {
    "Content-Type": "application/octet-stream",
    "apikey": "",  # coloque sua chave aqui
}
params = {
    "language": "pt",
}

for idx, row in df.iterrows():
    # Extrai só o nome do arquivo do caminho do CSV
    nome_arquivo = os.path.basename(row[0])
    # Monta o caminho correto
    image_path = os.path.join(base_dir, nome_arquivo)
    if not os.path.isfile(image_path):
        print(f"Arquivo NÃO encontrado: {image_path}")
        print("-" * 50)
        continue
    try:
        with open(image_path, "rb") as f:
            image_data = f.read()
        response = requests.post(
            base_url,
            headers=headers,
            params=params,
            data=image_data
        )
        print(f"Imagem {image_path}")
        print("Status code:", response.status_code)
        print("Resposta:", response.text)
        print("-" * 50)
    except Exception as e:
        print(f"Erro ao processar {image_path}: {e}")
        print("-" * 50)
```

# Ajustando parametros 

Exemplo está excelente para mostrar como as **tags** e a **descrição** podem ser usadas para customizar um modelo de visão computacional para o contexto da Petrobras.
Veja como ficaria o processo prático usando seu exemplo:
---
## **Exemplo prático de customização**
### 1. **Imagem e contexto**
- **Arquivo:** Dig39035.jpg
- **Descrição humana:** Homem trabalhando na sala de controle da Usina Termelétrica Jesus Soares Pereira
- **Tags do banco de imagens:**  
  - UTE_Jesus_Soares_Pereira  
  - Gás_e_Energia  
  - Empregados  
  - Empregados_Gas_e_Energia  
  - Energia_Elétrica  
  - Usina_Termeletrica
### 2. **Saída do modelo genérico**
- **Tags automáticas:** pessoa, no interior, teto, homem, em pé, frente, segurando, mesa, computador, laranja, tela, jovem, laptop, usando, quarto, vídeo, jogando, camisa, comida, pessoas, trem, jogo, tábua, remoto
- **Legenda automática:** homem em frente a computador (confiança: 66,5%)
### 3. **Como usar as tags e descrições para customizar**
- As **tags do banco de imagens** são muito mais específicas e relevantes para o contexto Petrobras do que as tags genéricas do modelo.
- Ao treinar ou ajustar um modelo, você pode usar essas tags como **rótulos de referência** (labels) para ensinar o modelo a reconhecer situações e ambientes industriais específicos.
---
## **Como fazer na prática**
1. **Monte um dataset**
   - Para cada imagem, associe as tags específicas do contexto Petrobras.
   - Exemplo de entrada para o modelo:
     - Imagem: Dig39035.jpg
     - Labels: [Usina_Termeletrica, Energia_Elétrica, Sala_de_Controle, Empregado, Gás_e_Energia]
2. **Treine o modelo**
   - Use frameworks como TensorFlow, PyTorch, ou plataformas como Azure Custom Vision, Google AutoML Vision.
   - Carregue as imagens e as tags como labels.
   - O modelo vai aprender a associar padrões visuais (ex: sala de controle, uniformes, monitores) com as tags específicas.
3. **Valide e ajuste**
   - Teste o modelo com novas imagens do mesmo contexto.
   - Veja se ele sugere as tags corretas (ex: “Usina_Termeletrica” ao invés de “quarto” ou “jogo”).
   - Se necessário, adicione mais imagens e refine as tags.
---
## **Dicas importantes**
- **Quanto mais imagens rotuladas, melhor o modelo aprende.**
- **Use sempre as tags do seu negócio, não as genéricas.**
- **Inclua descrições detalhadas para ajudar no treinamento supervisionado.**
- **Se possível, envolva especialistas do setor para validar as tags e descrições.**
---
## **Resumo**
Sim, você pode (e deve!) usar as tags específicas de cada imagem como referência para customizar o modelo. Isso vai garantir que ele reconheça corretamente ambientes, equipamentos e situações do setor industrial da Petrobras, ao invés de dar respostas genéricas.
