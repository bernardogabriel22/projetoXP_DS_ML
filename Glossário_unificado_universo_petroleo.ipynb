```python
# consolidação dos dados

import pandas as pd

# Arquivos, referência, separador e linhas a pular (para ignorar URL)
arquivos = [
    ("glossario_anp_processado.csv", "glossario_anp", ',', 0),
    ("glossario_anuario_petroleo.csv", "glossario_anuario_petroleo", ';', 0),
    ("glossario_petrobras_processado.csv", "glossario_petrobras", ',', 1),
    ("glossario_petroreconcavo_processado.csv", "glossario_petroreconcavo", ',', 1),
    ("glossario_presalpetroleo_processado.csv", "glossario_presalpetroleo", ',', 0)
]

dfs = []

for nome_arquivo, referencia, sep, skiprows in arquivos:
    print(f"Lendo arquivo: {nome_arquivo} (sep='{sep}', skiprows={skiprows})")
    try:
        df = pd.read_csv(nome_arquivo, sep=sep, skiprows=skiprows, encoding='utf-8')
    except Exception as e:
        print(f"Erro ao ler {nome_arquivo}: {e}")
        continue

    # Normaliza nomes das colunas: tira espaços, põe minúsculas e remove ':'
    df.columns = df.columns.str.strip().str.lower().str.replace(':', '')

    # Identifica colunas termo e definição (busca parcial)
    col_termo = next((c for c in df.columns if 'termo' in c), None)
    col_definicao = next((c for c in df.columns if 'defin' in c), None)

    if col_termo is None or col_definicao is None:
        print(f"Colunas 'termo' e 'definição' não encontradas no arquivo {nome_arquivo}")
        continue

    # Seleciona e renomeia colunas
    df = df[[col_termo, col_definicao]].copy()
    df.columns = ['TERMO', 'DEFINIÇÃO']

    # Remove linhas com termo ou definição vazios
    df = df.dropna(subset=['TERMO', 'DEFINIÇÃO'])

    # Adiciona coluna de referência
    df['REFERENCIA'] = referencia

    dfs.append(df)

if dfs:
    df_final = pd.concat(dfs, ignore_index=True)
    df_final.to_csv('glossario_consolidado.csv', index=False, sep=';', encoding='utf-8-sig')
    print("CSV consolidado criado com sucesso!")
else:
    print("Nenhum dado foi carregado.")
```

```python


import pandas as pd
from IPython.display import display
import ipywidgets as widgets

# Carrega o DataFrame
df = pd.read_csv('glossario_consolidado.csv', sep=';', encoding='utf-8-sig')

# Widgets de texto para filtro
filtro_termo = widgets.Text(
    value='',
    placeholder='Filtrar termo',
    description='Termo:',
    disabled=False
)

filtro_definicao = widgets.Text(
    value='',
    placeholder='Filtrar definição',
    description='Definição:',
    disabled=False
)

# Função para atualizar a visualização filtrada
def filtrar_df(change):
    termo_valor = filtro_termo.value.lower()
    definicao_valor = filtro_definicao.value.lower()

    df_filtrado = df

    if termo_valor:
        df_filtrado = df_filtrado[df_filtrado['TERMO'].str.lower().str.contains(termo_valor)]
    if definicao_valor:
        df_filtrado = df_filtrado[df_filtrado['DEFINIÇÃO'].str.lower().str.contains(definicao_valor)]

    display(df_filtrado[['TERMO', 'DEFINIÇÃO', 'REFERENCIA']].reset_index(drop=True))

# Widget interativo
out = widgets.Output()

def on_text_change(change):
    out.clear_output()
    with out:
        filtrar_df(change)

filtro_termo.observe(on_text_change, names='value')
filtro_definicao.observe(on_text_change, names='value')

display(filtro_termo, filtro_definicao, out)

# Exibe inicialmente tudo
with out:
    display(df[['TERMO', 'DEFINIÇÃO', 'REFERENCIA']].head(20))

```

```python
#3. Pré-processamento básico das definições
#Exemplo para tokenizar, remover stopwords e lematizar:

import spacy

nlp = spacy.load("pt_core_news_sm")  # modelo em português

def preprocessar(texto):
    doc = nlp(str(texto).lower())
    tokens = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]
    return tokens

df['definicao_tokens'] = df['DEFINIÇÃO'].apply(preprocessar)

df
```

Obs: Estruturas matriciais são modelos de organização que combinam departamentalização funcional com a divisão por projeto ou produto, resultando em uma dupla subordinação para os colaboradores, que se reportam a um gerente funcional e a um gerente de projeto simultaneamente. Essa estrutura promove a colaboração entre áreas, utiliza recursos de forma eficiente e aumenta a flexibilidade para lidar com projetos complexos


### Características principais

Dupla autoridade: O ponto central da estrutura matricial é que os funcionários têm dois chefes: um gerente funcional (responsável pela área de especialidade, como marketing ou engenharia) e um gerente de projeto (responsável por um projeto específico).

Equipes multidisciplinares: Profissionais de diferentes departamentos são alocados para trabalhar juntos em projetos específicos, compartilhando conhecimentos e habilidades.

Flexibilidade: É ideal para empresas que precisam gerenciar múltiplos projetos ao mesmo tempo, pois permite a rápida realocação de recursos e equipes para diferentes iniciativas.

### Vantagens

Otimização de recursos: Permite o uso mais eficiente de talentos e habilidades especializadas, já que o mesmo profissional pode trabalhar em diferentes projetos.

Melhora da comunicação: A comunicação entre os diferentes setores é fortalecida, pois as equipes se reúnem para trabalhar em objetivos comuns de projeto.

Desenvolvimento de colaboradores: Oferece aos funcionários a oportunidade de trabalhar em projetos variados, o que pode aumentar o engajamento e expandir suas habilidades.

### Desvantagens

Potencial para conflitos: A dupla subordinação pode gerar conflitos entre os gerentes sobre prioridades e recursos, e os funcionários podem ficar confusos sobre a quem devem obedecer.

Necessidade de comunicação clara: Requer uma comunicação constante e clara entre os gerentes e a equipe para garantir o alinhamento e evitar atritos.
Complexidade: É uma estrutura mais complexa do que as tradicionais, exigindo maturidade organizacional e líderes capacitados para gerenciar a dualidade de comando.
